```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# ==========================================
# File paths
# ==========================================

input_path = Path(
    "../data/raw/lahore/lahore_weather_pollution_merged_hourly.csv"
)

output_path = Path(
    "../data/processed/lahore/lahore_cleaned.csv"
)

# Create processed directory if it doesn't exist
output_path.parent.mkdir(parents=True, exist_ok=True)

print("Input:", input_path)
print("Output:", output_path)
```

```python
# ==========================================
# Load raw dataset
# ==========================================

lahore_df = pd.read_csv(input_path)

print("Dataset loaded successfully.")
print("Shape:", lahore_df.shape)
```

```python
print("Columns:")
print(lahore_df.columns.tolist())
```

```python
print(lahore_df.dtypes)
```

```python
# ==========================================
# Convert datetime
# ==========================================

lahore_df["datetime"] = pd.to_datetime(
    lahore_df["datetime"],
    errors="coerce"
)

print(lahore_df["datetime"].dtype)
```

```python
invalid_datetime = lahore_df["datetime"].isna().sum()

print("Invalid datetime values:", invalid_datetime)
```

```python
lahore_df = (
    lahore_df
    .sort_values("datetime")
    .reset_index(drop=True)
)

print("Data sorted by datetime.")
```

```python
duplicate_timestamps = lahore_df["datetime"].duplicated().sum()

print("Duplicate timestamps:", duplicate_timestamps)
```

```python
missing_values = lahore_df.isnull().sum()

print("Missing values:")
print(missing_values)
```

```python
print(
    "Total missing values:",
    lahore_df.isnull().sum().sum()
)
```

```python
numeric_columns = lahore_df.select_dtypes(
    include=np.number
).columns

print("Numeric columns:")
print(numeric_columns.tolist())
```

```python
lahore_df[numeric_columns].describe().T
```

```python
non_negative_columns = [
    "temperature_2m",
    "relative_humidity_2m",
    "surface_pressure",
    "precipitation",
    "cloud_cover",
    "wind_speed_10m",
    "pm2_5",
    "pm10",
    "carbon_monoxide",
    "nitrogen_dioxide",
    "sulphur_dioxide",
    "ozone",
    "us_aqi"
]

for column in non_negative_columns:
    negative_count = (lahore_df[column] < 0).sum()
    print(f"{column}: {negative_count} negative values")
```

```python
print(
    "Relative humidity range:",
    lahore_df["relative_humidity_2m"].min(),
    "to",
    lahore_df["relative_humidity_2m"].max()
)

print(
    "Cloud cover range:",
    lahore_df["cloud_cover"].min(),
    "to",
    lahore_df["cloud_cover"].max()
)

print(
    "Wind direction range:",
    lahore_df["wind_direction_10m"].min(),
    "to",
    lahore_df["wind_direction_10m"].max()
)

print(
    "AQI range:",
    lahore_df["us_aqi"].min(),
    "to",
    lahore_df["us_aqi"].max()
)
```

```python
time_difference = lahore_df["datetime"].diff()

print(
    "Most common time interval:"
)

print(
    time_difference.value_counts().head()
)
```

```python
expected_interval = pd.Timedelta(hours=1)

gaps = lahore_df[
    time_difference != expected_interval
]

print("Number of non-hourly gaps:", len(gaps))
```

```python
print("Minimum datetime:", lahore_df["datetime"].min())
print("Maximum datetime:", lahore_df["datetime"].max())
```

```python
print("===================================")
print("FINAL DATA VALIDATION")
print("===================================")

print("Rows:", len(lahore_df))
print("Columns:", len(lahore_df.columns))
print("Missing values:", lahore_df.isnull().sum().sum())
print("Duplicate timestamps:", lahore_df["datetime"].duplicated().sum())
print("Invalid datetime:", lahore_df["datetime"].isna().sum())
print("Minimum datetime:", lahore_df["datetime"].min())
print("Maximum datetime:", lahore_df["datetime"].max())
```

```python
# ==========================================
# Check missing values
# ==========================================

print("Missing values:")
print(lahore_df.isnull().sum())
```

```python
# ==========================================
#  Check duplicate timestamps
# ==========================================

duplicates = lahore_df["datetime"].duplicated().sum()

print("Duplicate timestamps:", duplicates)
```

```python
# ==========================================
#  Check hourly continuity
# ==========================================

time_diff = lahore_df["datetime"].diff()

non_hourly = time_diff[
    time_diff.notna() &
    (time_diff != pd.Timedelta(hours=1))
]

print("Number of non-hourly gaps:", len(non_hourly))

if len(non_hourly) > 0:
    print("\nGap details:")

    gap_indices = non_hourly.index

    for idx in gap_indices:
        previous_time = lahore_df.loc[idx - 1, "datetime"]
        current_time = lahore_df.loc[idx, "datetime"]

        print(
            f"{previous_time} → {current_time} "
            f"= {current_time - previous_time}"
        )
```

```python
# ==========================================
# Save cleaned dataset
# ==========================================

lahore_df.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print(output_path)
```
